# What is the total revenue generated from all sales?
# Insight:
The query calculates the total revenue generated from all sales transactions in the dataset.

```python
query="""
SELECT SUM(Total_Sales) AS Total_Revenue
FROM sales_data;
"""
result=pd.read_sql(query,engine)
result
```

# How many total orders are present?

```python
query="""
SELECT COUNT(*) AS Total_Orders
FROM sales_data;
"""
result=pd.read_sql(query,engine)
result
```

# Which product category generates the highest revenue?

```python
query="""
SELECT SUM(Total_Sales) AS Total_Revenue
FROM sales_data
GROUP BY Category
ORDER BY Total_Revenue DESC;
"""

result=pd.read_sql(query,engine)
result
```

# What are the top 5 products by revenue?


```python
query="""
SELECT Product,SUM(Total_Sales) AS Total_Revenue
FROM sales_data
GROUP BY Product
ORDER BY Total_Revenue DESC
LIMIT 5;
"""
result=pd.read_sql(query,engine)
result
```

# What is the monthly revenue trend?

```python
query="""
SELECT AVG(Total_Sales) AS AVG_MOMTHLY_TREND
FROM sales_data;
"""
result=pd.read_sql(query,engine)
result
```

# Which cities generate the highest revenue?

```python
query="""
SELECT City,SUM(Total_Sales) AS Total_Revenue
FROM sales_data
GROUP BY City
ORDER BY Total_Revenue DESC;
"""
result=pd.read_sql(query,engine)
result
```

# Correlation Matrix

```python
import seaborn as sns

corr=df[["Age","Quantity","Unit_Price","Total_Sales","Order_Month"]].corr()
corr
```

# Correlation Heatmap:-

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f")

plt.title("correction heatmap")
plt.show()
```

# Quantity vs Total Sales:-

```python
plt.figure(figsize=(8,6))
plt.scatter(df["Quantity"],df["Total_Sales"],alpha=0.6)

plt.xlabel("Quantity")
plt.ylabel("Total_Sales")
plt.title("Quantity Vs Total_Sales")

plt.show()
```

# Unit Price vs Total Sales:-

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Unit_Price"],df["Total_Sales"],alpha=0.6)

plt.title("Unit_Price Vs Total_Sales")
plt.xlabel("Unit_Price")
plt.ylabel("Total_Price")
plt.show()
```

# Age vs Total Sales:-

```python
plt.figure(figsize=(8,6))
plt.scatter(df["Age"],df["Total_Sales"],alpha=0.6)
plt.title("Age Vs Total_Price")
plt.xlabel("Age")
plt.ylabel("Total_Price")
plt.show()
```
